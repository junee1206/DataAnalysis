```python
import numpy as np
import pandas as pd
%matplotlib inline
import matplotlib.pyplot as plt
```

```python
plt.plot()
plt.show()
```

```python
x = [1,2,3,4]
y =[4,3,2,1]

plt.plot(x,y)
plt.show()
```

```python
# recommended method 
fig,ax = plt.subplots()
ax.plot(x,y);
```

```python
# Matplotlib workflow 
fig, ax = plt.subplots(figsize=(6,6))
ax.plot(x,y)
ax.set(title="Simple Plot",
       xlabel="X Axis",
       ylabel="Y Axis")
```

```python
x = np.linspace(0,10,100)
x[:10]
```

```python
# Line Plot 
fig, ax = plt.subplots(figsize=(8,5))
ax.plot(x,x**2)
```

```python
fig,ax  = plt.subplots()
ax.scatter(x,x**2)
```

```python
fig,ax  = plt.subplots()
ax.scatter(x,np.sin(x))
```

```python
nut_butter_price = {
    "Almond Butter":10,
    "Peanut Butter":8,
    "Cashew Butter": 12
}

fig, ax = plt.subplots()
ax.bar(nut_butter_price.keys(), nut_butter_price.values())
```

```python
fig, ax = plt.subplots()
ax.barh(nut_butter_price.keys(), nut_butter_price.values())
```

```python
x = np.random.randn(100)
fig,ax = plt.subplots()
ax.hist(x);
```

```python
# Subplot 

fig, (ax1,ax2) = plt.subplots(nrows=1,ncols=2,figsize=(10,5))
ax1.plot(x,x/2)
ax2.plot(x,np.exp(x))
```

```python
# Subplot 

fig, ax = plt.subplots(nrows=1,ncols=2,figsize=(10,5))
ax[0].plot(x,x/2)
ax[1].plot(x,np.exp(x))
```

```python
car_sales = pd.read_csv('car-sales.csv')
car_sales
```

```python
car_sales["Price"] = car_sales["Price"].str.replace(r'[\$,]', '', regex=True).astype(float).astype(int)
```

```python
car_sales
```

```python
car_sales["Date"] = pd.date_range("01/01/2026",periods=len(car_sales))
car_sales
```

```python
car_sales["Total Sales"] = car_sales["Price"].cumsum()
car_sales
```

```python
car_sales.plot("Date","Total Sales")
```

```python
car_sales.plot("Odometer (KM)","Price",kind="scatter")
```

```python
car_sales.plot("Make","Price",kind="bar")
```
