# 7. Filtering Data ⭐⭐⭐

### One condition

```python
df[df["Age"] > 25]
```

### Multiple conditions

```python
df[(df["Age"] > 25) & (df["Salary"] > 40000)]
```

OR:

```python
df[(df["Age"] > 25) | (df["Salary"] > 40000)]
```

NOT:

```python
df[~(df["Age"] > 25)]
```

### `isin()`

```python
df[df["City"].isin(["Delhi", "Mumbai"])]
```

### String filtering

```python
df[df["Name"].str.contains("John", na=False)]
```

# 34. Find Outliers Quickly

Using quantiles:

```python
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

IQR = Q3 - Q1

outliers = df[
    (df["Salary"] < Q1 - 1.5 * IQR) |
    (df["Salary"] > Q3 + 1.5 * IQR)
]
```

# 36. Common Data Cleaning Pipeline ⭐⭐⭐⭐⭐

A typical Data Science workflow:

```python
import pandas as pd

df = pd.read_csv("data.csv")

# Inspect
df.head()
df.info()
df.describe()

# Remove duplicates
df = df.drop_duplicates()

# Missing values
df["Age"] = df["Age"].fillna(
    df["Age"].median()
)

# Convert datatype
df["Salary"] = pd.to_numeric(
    df["Salary"],
    errors="coerce"
)

# Clean strings
df["Name"] = df["Name"].str.strip()

# Feature creation
df["Annual_Salary"] = df["Monthly_Salary"] * 12

# Filtering
df = df[df["Age"] >= 18]

# Reset index
df = df.reset_index(drop=True)
```
