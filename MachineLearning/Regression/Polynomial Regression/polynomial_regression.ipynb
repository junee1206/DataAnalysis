```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv('Position_Salaries.csv')
dataset.head()
```

```python
X = dataset.iloc[:, 1:-1].values
y = dataset.iloc[:, -1].values
```

```python
X
```

```python
y
```

```python
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(X,y)
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly_regressor = PolynomialFeatures(degree=2)
X_poly = poly_regressor.fit_transform(X)

lin_regressor  = LinearRegression()
lin_regressor.fit(X_poly,y)
```

```python
plt.scatter(X,y,color='red')
plt.plot(X,regressor.predict(X),color='blue')
plt.title('Truth or Bluff (Linear Regression)')
plt.xlabel('Position Level')
plt.ylabel('Salary')
plt.show()
```

```python
plt.scatter(X,y,color='red')
plt.plot(X,lin_regressor.predict(X_poly),color='blue')
plt.title('Truth or Bluff (Polynomial Regression)')
plt.xlabel('Position Level')
plt.ylabel('Salary')
plt.show()
```

```python
poly_regressor = PolynomialFeatures(degree=3)
X_poly1 = poly_regressor.fit_transform(X)

lin_regressor1  = LinearRegression()
lin_regressor1.fit(X_poly1,y)
```

```python
plt.scatter(X,y,color='red')
plt.plot(X,lin_regressor1.predict(X_poly1),color='blue')
plt.title('Truth or Bluff (Polynomial Regression  Degree 3)')
plt.xlabel('Position Level')
plt.ylabel('Salary')
plt.show()
```

```python
poly_regressor = PolynomialFeatures(degree=4)
X_poly2 = poly_regressor.fit_transform(X)

lin_regressor2  = LinearRegression()
lin_regressor2.fit(X_poly2,y)
```

```python
plt.scatter(X,y,color='red')
plt.plot(X,lin_regressor2.predict(X_poly2),color='blue')
plt.title('Truth or Bluff (Polynomial Regression  Degree 4)')
plt.xlabel('Position Level')
plt.ylabel('Salary')
plt.show()
```
